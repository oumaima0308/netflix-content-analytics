```python
from urllib.parse import quote_plus
from sqlalchemy import create_engine
```

```python
password = "YourPassword"

encoded_password = quote_plus(password)

engine = create_engine(
    f"mysql+pymysql://root:{encoded_password}@localhost:3306/netflix_bi"
)
```

```python
with engine.connect() as connection:
    print("Connexion MySQL réussie !")
```

```python
countries.to_sql(
    "countries",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
categories.to_sql(
    "categories",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
content.to_sql(
    "content",
    con=engine,
    if_exists="append",
    index=False
)
```
